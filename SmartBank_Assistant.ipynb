```python
import os
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity

# Emplacement des documents
documents_dir = r"C:\Users\ACER\SmartBank_Assistant\documents"

# Lire les documents
documents = []

for filename in os.listdir(documents_dir):
    if filename.endswith(".txt"):
        path = os.path.join(documents_dir, filename)

        with open(path, "r", encoding="utf-8") as file:
            content = file.read()

        documents.append({
            "nom": filename,
            "contenu": content
        })

# Transformer les documents en représentations numériques
vectorizer = TfidfVectorizer()
textes = [doc["contenu"] for doc in documents]
matrice = vectorizer.fit_transform(textes)

print("Moteur de recherche documentaire créé !")
print("Nombre de documents chargés :", len(documents))
```

```python
def rechercher_information(question):
    # Transformer la question en représentation numérique
    question_vecteur = vectorizer.transform([question])

    # Calculer la similarité avec les documents
    scores = cosine_similarity(question_vecteur, matrice)[0]

    # Trouver le document le plus pertinent
    meilleur_index = scores.argmax()

    # Vérifier si le résultat est suffisamment pertinent
    if scores[meilleur_index] == 0:
        return "Je ne trouve pas cette information dans les documents SmartBank."

    return documents[meilleur_index]["contenu"]


# Premier test
question = "Combien coûte une carte bancaire classique ?"

reponse = rechercher_information(question)

print("Question du client :", question)
print("\nRéponse retrouvée :")
print(reponse)
```

```python

import re
from sklearn.metrics.pairwise import cosine_similarity

def chatbot_conseiller(question):

    # Découper les documents en phrases
    passages = []

    for doc in documents:
        phrases = re.split(r'(?<=[.!?])\s+|\n+', doc["contenu"])

        for phrase in phrases:
            phrase = phrase.strip()

            if len(phrase) > 5:
                passages.append({
                    "texte": phrase,
                    "document": doc["nom"]
                })

    # Transformer les passages en vecteurs
    textes = [p["texte"] for p in passages]
    vecteurs = vectorizer.transform(textes)

    # Comparer la question aux passages
    question_vecteur = vectorizer.transform([question])
    scores = cosine_similarity(question_vecteur, vecteurs)[0]

    # Sélectionner le passage le plus pertinent
    meilleur_index = scores.argmax()
    meilleur_score = scores[meilleur_index]

    if meilleur_score == 0:
        return "Je suis désolé, je ne trouve pas cette information dans les documents SmartBank."

    meilleur_passage = passages[meilleur_index]["texte"]

    return "Voici l'information trouvée : " + meilleur_passage


# Tester le chatbot
question = "Combien coûte une carte bancaire classique ?"

print("Client :", question)
print("SmartBank Assistant :", chatbot_conseiller(question))
```

```python

import re
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity

# Regrouper les lignes liées dans des paragraphes
passages = []

for doc in documents:
    paragraphes = re.split(r"\n\s*\n", doc["contenu"])

    for paragraphe in paragraphes:
        paragraphe = " ".join(paragraphe.split())

        if len(paragraphe) > 10:
            passages.append({
                "texte": paragraphe,
                "document": doc["nom"]
            })

# Créer un nouveau moteur adapté aux paragraphes
textes_passages = [p["texte"] for p in passages]

vectorizer_chatbot = TfidfVectorizer(
    ngram_range=(1, 2),
    strip_accents="unicode"
)

matrice_chatbot = vectorizer_chatbot.fit_transform(textes_passages)


def chatbot_conseiller(question):

    # Quelques adaptations pour les formulations courantes
    question_corrigee = question.lower()
    question_corrigee = question_corrigee.replace("perds", "perte")
    question_corrigee = question_corrigee.replace("perdu", "perte")

    # Comparer la question aux paragraphes
    vecteur_question = vectorizer_chatbot.transform(
        [question_corrigee]
    )

    scores = cosine_similarity(
        vecteur_question, matrice_chatbot
    )[0]

    meilleur_index = scores.argmax()
    meilleur_score = scores[meilleur_index]

    if meilleur_score < 0.08:
        return "Je ne trouve pas cette information dans les documents SmartBank."

    resultat = passages[meilleur_index]

    return (
        "Information trouvée dans " + resultat["document"] +
        " :\n" + resultat["texte"]
    )


print("Nouveau moteur de recherche créé !")
print("Nombre de passages :", len(passages))
```

```python

import os

app_path = r"C:\Users\ACER\SmartBank_Assistant\app.py"

with open(app_path, "r", encoding="utf-8") as file:
    code = file.read()

debut = code.index("def rechercher_reponse(question):")
fin = code.index("# Navigation", debut)

nouvelle_fonction = '''
def rechercher_reponse(question):

    import unicodedata
    import re

    # Normalisation de la question
    q = question.lower()
    q = unicodedata.normalize("NFD", q)
    q = "".join(c for c in q if unicodedata.category(c) != "Mn")
    q = re.sub(r"[^a-z0-9 ]", " ", q)
    q = " ".join(q.split())

    # 1. PERTE OU VOL DE CARTE
    if any(x in q for x in [
        "perdu ma carte", "perte de carte", "perds ma carte",
        "vole ma carte", "vol de carte", "carte disparue",
        "opposition a ma carte", "faire opposition"
    ]):
        return (
            "En cas de perte ou de vol de votre carte, "
            "contactez immédiatement le service d'opposition."
        )

    # 2. VIREMENT NON RECU
    if (
        any(x in q for x in ["virement", "transfert", "transfert bancaire"])
        and any(x in q for x in [
            "pas recu", "non recu", "pas arrive", "non arrive",
            "toujours pas", "pas encore", "en attente",
            "manquant", "recu", "arrive"
        ])
    ):
        return (
            "Pour signaler un virement non reçu, veuillez fournir "
            "la date, le montant et la référence du virement."
        )

    # 3. RECLAMATION
    if any(x in q for x in [
        "reclamation", "reclamer", "plainte",
        "signaler un probleme", "probleme bancaire"
    ]):
        return (
            "Pour déposer une réclamation, indiquez le motif, "
            "la date du problème et joignez les justificatifs disponibles."
        )

    # 4. CARTE BANCAIRE CLASSIQUE
    if any(x in q for x in [
        "carte classique", "carte bancaire classique",
        "prix carte classique", "cout carte classique"
    ]):
        return "La carte bancaire classique coûte 150 DH par an."

    # 5. CARTE BANCAIRE PREMIUM
    if any(x in q for x in [
        "carte premium", "carte bancaire premium",
        "prix carte premium", "cout carte premium"
    ]):
        return "La carte bancaire premium coûte 350 DH par an."

    # 6. TENUE DE COMPTE
    if any(x in q for x in [
        "tenue de compte", "frais de compte",
        "frais compte", "cout du compte"
    ]):
        return "Les frais annuels de tenue de compte sont de 120 DH."

    # 7. VIREMENT NATIONAL EN LIGNE
    if any(x in q for x in [
        "virement national", "virement en ligne",
        "transfert national", "transfert en ligne"
    ]):
        return "Un virement national en ligne coûte 5 DH."

    # 8. VIREMENT EN AGENCE
    if any(x in q for x in [
        "virement en agence", "virement agence",
        "transfert en agence"
    ]):
        return "Un virement effectué en agence coûte 15 DH."

    # 9. RETRAIT SMARTBANK
    if any(x in q for x in [
        "retrait smartbank", "distributeur smartbank",
        "retrait dans votre banque", "retrait distributeur smartbank"
    ]):
        return "Les retraits aux distributeurs SmartBank sont gratuits."

    # 10. RETRAIT AUTRE BANQUE
    if any(x in q for x in [
        "autre banque", "autre distributeur",
        "distributeur d une autre banque",
        "retrait hors smartbank"
    ]):
        return "Un retrait dans une autre banque coûte 6 DH."

    # 11. CREDIT IMMOBILIER : TAUX
    if (
        "immobilier" in q
        and any(x in q for x in ["taux", "interet", "pourcentage"])
    ):
        return "Le taux annuel indicatif du crédit immobilier est de 4,5 %."

    # 12. CREDIT IMMOBILIER : DUREE
    if (
        "immobilier" in q
        and any(x in q for x in ["duree", "longtemps", "annees", "ans"])
    ):
        return "La durée maximale du crédit immobilier est de 25 ans."

    # 13. CREDIT IMMOBILIER : APPORT
    if (
        "immobilier" in q
        and any(x in q for x in ["apport", "personnel", "financement"])
    ):
        return "L'apport personnel indicatif pour un crédit immobilier est de 20 %."

    # 14. CREDIT A LA CONSOMMATION : TAUX
    if (
        any(x in q for x in ["consommation", "consommateur"])
        and any(x in q for x in ["taux", "interet", "pourcentage"])
    ):
        return "Le taux annuel indicatif du crédit à la consommation est de 6,5 %."

    # 15. CREDIT A LA CONSOMMATION : DUREE
    if (
        any(x in q for x in ["consommation", "consommateur"])
        and any(x in q for x in ["duree", "longtemps", "annees", "ans"])
    ):
        return "La durée maximale du crédit à la consommation est de 7 ans."

    # 16. CREDIT A LA CONSOMMATION : MONTANT
    if (
        any(x in q for x in ["consommation", "consommateur"])
        and any(x in q for x in ["montant", "maximum", "maximal", "combien"])
    ):
        return "Le montant maximal du crédit à la consommation est de 150 000 DH."

    # 17. CONDITIONS GENERALES DU CREDIT
    if "credit" in q and any(x in q for x in [
        "condition", "obtenir", "demander", "eligibilite"
    ]):
        return (
            "Pour demander un crédit, vous devez déposer un dossier de financement. "
            "Les conditions définitives dépendent de l'étude du dossier "
            "et de la décision de la banque."
        )

    # 18. OUVERTURE DE COMPTE
    if any(x in q for x in [
        "ouvrir un compte", "ouverture de compte",
        "ouvrir compte", "documents compte"
    ]):
        return (
            "Pour ouvrir un compte, présentez une pièce d'identité, "
            "un justificatif de domicile et les documents demandés par la banque."
        )

    # 19. OBTENIR UNE CARTE
    if any(x in q for x in [
        "obtenir une carte", "commander une carte",
        "demander une carte", "avoir une carte"
    ]):
        return "Pour obtenir une carte bancaire, veuillez contacter un conseiller."

    # 20. CONTACTER UN CONSEILLER
    if any(x in q for x in [
        "contacter un conseiller", "parler a un conseiller",
        "joindre un conseiller", "contact conseiller"
    ]):
        return "Pour obtenir une assistance, veuillez contacter un conseiller SmartBank."

    # 21. RENDEZ-VOUS
    if any(x in q for x in [
        "rendez vous", "prendre rendez vous",
        "rdv", "rencontrer un conseiller"
    ]):
        return "Vous pouvez prendre rendez-vous avec un conseiller depuis le service de rendez-vous SmartBank."

    # 22. DOCUMENTS POUR UN CREDIT
    if "credit" in q and any(x in q for x in [
        "document", "piece", "dossier", "justificatif"
    ]):
        return (
            "Pour demander un crédit, vous devez déposer un dossier de financement. "
            "Les documents précis dépendent de votre situation et de l'étude du dossier."
        )

    # 23. QUESTIONS SUR LE CREDIT IMMOBILIER
    if "maison" in q or "acheter un logement" in q:
        return (
            "Pour un financement immobilier, la durée maximale indicative est de 25 ans, "
            "l'apport personnel indicatif est de 20 % et le taux annuel indicatif est de 4,5 %. "
            "L'accord dépend de l'étude de votre dossier."
        )

    # 24. QUESTION SUR LE CREDIT EN GENERAL
    if "credit" in q:
        return (
            "SmartBank propose un crédit immobilier et un crédit à la consommation. "
            "Les conditions dépendent du type de crédit et de l'étude de votre dossier."
        )

    # 25. QUESTION SUR LES CARTES EN GENERAL
    if "carte" in q:
        return (
            "SmartBank propose une carte classique à 150 DH par an "
            "et une carte premium à 350 DH par an."
        )

    # Réponse prudente si aucune intention n'est reconnue
    return (
        "Je suis désolé, je ne trouve pas une réponse suffisamment précise "
        "dans les documents SmartBank. Veuillez reformuler votre question "
        "ou contacter un conseiller."
    )
'''

code = code[:debut] + nouvelle_fonction + code[fin:]

with open(app_path, "w", encoding="utf-8") as file:
    file.write(code)

print("Le nouveau système de réponses a été installé avec succès !")
'''

with open(app_path, "w", encoding="utf-8") as file:
    file.write(code)

print("Le nouveau système de réponses a été installé avec succès !")
```
